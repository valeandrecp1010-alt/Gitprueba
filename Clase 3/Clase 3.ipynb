```python
lista_num = [3.5, 1, 75, 2.43, 56, 32]
lista_num.append(43.6)
print(lista_num)
```

```python
#Ordenar de menor a mayor
lista_num.sort()
print(lista_num)
```

```python
#Orden de mayor a menor
lista_num.sort(reverse = True)
print(lista_num)
```

```python
#Eliminar un dato
lista_num.pop(-3)
print(lista_num)
```

```python
#Maximo, minimo, suma, cantidad
print("El maximo es: ", max(lista_num))
print("El minimo es: ", min(lista_num))
print("La suma es: ", sum(lista_num))
print("la cantidad es: ", len(lista_num))
```
