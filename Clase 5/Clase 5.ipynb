```python
notas_arreglo = np.array([3.2, 4.5, 2.8, 3.9, 2.1])
print(notas_arreglo >= 3)
```

```python
notas_arreglo1 = np.array([3.2, 4.5, 2.8, 3.9, 2.1, 3.2, 4.5])

print("El promedio es: ", notas_arreglo1.mean().round(2))
print("El mediana es: ", np.median(notas_arreglo1))
print("La suma de los valores es: ", notas_arreglo1.sum())
print("El maximo es: ", notas_arreglo1.max())
print("El minimo es: ", notas_arreglo1.min())
print("La desviacion estandar es: ", notas_arreglo1.std().round(2))
print("Los valores unicos son: ", np.unique(notas_arreglo1))
```
