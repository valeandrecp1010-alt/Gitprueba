```python
edad1 = int(input("Ingrese la edad 1: "))
edad2 = int(input("Ingrese la edad 2: "))
edad3 = int(input("Ingrese la edad 3: "))
edad4 = int(input("Ingrese la edad 4: "))
edad5 = int(input("Ingrese la edad 5: "))
edad6 = int(input("Ingrese la edad 6: "))
edad7 = int(input("Ingrese la edad 7: "))
edad8 = int(input("Ingrese la edad 8: "))
edad9 = int(input("Ingrese la edad 9: "))
edad10 = int(input("Ingrese la edad 10: "))

promedio = (edad1+edad2+edad3+edad4+edad5+edad6+edad7+edad8+edad9+edad10)/10
print("El promedio es: ", promedio)
```

```python
lista_edades = []
for i in range(10):
  edad = int(input(f"Ingrese la edad {i+1}: "))
  lista_edades.append(edad)
promedio = sum(lista_edades)/len(lista_edades)
print("El proimedio es: ", promedio)
```
