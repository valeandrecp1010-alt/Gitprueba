```python
from matplotlib.lines import lineStyles
import numpy as np
import matplotlib.pyplot as plt

x = np.arange(5)
y = x**2

plt.plot(x, y, "r--o", label = "lineal")
plt.plot(x, y, color = "green", linestyle = ":", marker = "s", linewidth = "2", label = "Cuadratica")
plt.title("Grafico de curvas, lineas y cuadratica: ")
plt.xlabel("Eje x")
plt.ylabel("Eje y")
plt.xlim(0,6)
plt.ylim(0,6)
plt.legend()
plt.grid()
plt.show()
```

```python
plt.scatter(x,y, label = "Dispersión")
plt.title("Grafico de dispersión")
plt.xlabel("Eje x")
plt.ylabel("Eje y")
plt.legend()
plt.grid()
plt.show()
```

```python
equipos = ["Celular", "Parlantes", "Audifonos", "Pc", "Portatil"]
cantidad = [3, 6, 2, 2, 4]

plt.bar(equipos, cantidad)
plt.title("Grafico de barras: ")
plt.xlabel("Equipos")
plt.ylabel("Cantidad")
plt.grid()
plt.show()
```

```python
fuentes = ["Hidráulica", "Termica", "Solar", "Eólica"]
porc = [55, 30, 10, 5]
plt.pie(porc, labels=fuentes, autopct="%1.0f%%")
plt.title("Generación (ejemplo)")
plt.show()
```
